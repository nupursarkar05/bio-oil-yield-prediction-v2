```python
def clean_numeric(value):
    """
    Convert messy numeric strings into floats.
    """

    if pd.isna(value):
        return np.nan

    value = str(value).strip()
    value = value.replace(" ", "")

    try:
        if "-" in value:
            low, high = value.split("-")
            return (float(low) + float(high)) / 2

        return float(value)

    except:
        return np.nan
```

```python
columns = [
    "Particle_Size",
    "Biochar_Yield",
    "Bio_Oil_Yield",
    "Gas_Yield"
]

for col in columns:
    df_clean[col] = df_clean[col].apply(clean_numeric)
```
